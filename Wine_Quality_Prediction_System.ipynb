```python
# Split the dataset into 80% training and 20% testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

Observation — Quality Class Distribution

The quality-class proportions in the training and testing sets are very similar. This shows that the stratified split has successfully maintained the distribution of the quality classes in both datasets.

#Task 8 — Feature Scaling

```python
# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()
```

```python
# Fit the scaler using training data
scaler.fit(X_train)
```

```python
# Transform the training and testing features
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Display the first 5 scaled training samples
X_train_scaled[:5]
```

```python
# Check the mean of scaled training features
print("Mean of scaled features:")
print(X_train_scaled.mean(axis=0))

# Check the standard deviation
print("\nStandard deviation of scaled features:")
print(X_train_scaled.std(axis=0))
```

###Scaling Sensitivity of Models

Logistic Regression, Support Vector Machine (SVM), and K-Nearest Neighbors (KNN) are sensitive to differences in feature scale. Random Forest is generally less affected by feature scaling because it is a tree-based model

###Why Scaling is Important

The wine dataset contains numerical features with different ranges. Feature scaling brings these variables to a comparable scale, which is especially important for distance-based and margin-based algorithms such as KNN and SVM. This helps prevent features with larger numerical ranges from having an undue influence on the model.

##Task 8 - Final Observation
The scaled feature values are centered around 0, and the features are brought to a comparable scale. This reduces the effect of differences in the original feature ranges and makes the data more suitable for scaling-sensitive models such as Logistic Regression, SVM, and KNN

###Task 9 - Model Training

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Create the model
logistic_model = LogisticRegression(max_iter=1000)

# Train the model
logistic_model.fit(X_train_scaled, y_train)

# Generate predictions
y_pred_logistic = logistic_model.predict(X_test_scaled)

print("Logistic Regression predictions:")
print(y_pred_logistic[:10])
```

Why this model is suitable

Logistic Regression is suitable for this project because wine quality is treated as a classification problem. It provides a simple baseline model for predicting the different quality classes.

```python
# Import Random Forest
from sklearn.ensemble import RandomForestClassifier

# Create the model
random_forest_model = RandomForestClassifier(
    random_state=42
)

# Train the model
random_forest_model.fit(X_train_scaled, y_train)

# Generate predictions
y_pred_rf = random_forest_model.predict(X_test_scaled)

print("Random Forest predictions:")
print(y_pred_rf[:10])
```

Why this model is suitable

Random Forest is suitable for this classification problem because it combines multiple decision trees to make predictions and can capture complex relationships between the wine features and quality classes.

```python
# Import Support Vector Machine
from sklearn.svm import SVC

# Create the model
svm_model = SVC()

# Train the model
svm_model.fit(X_train_scaled, y_train)

# Generate predictions
y_pred_svm = svm_model.predict(X_test_scaled)

print("SVM predictions:")
print(y_pred_svm[:10])
```

Why this model is suitable

Support Vector Machine is suitable for this project because it is a classification algorithm that can find decision boundaries between different wine quality classes. Feature scaling is particularly important for SVM.

```python
# Import K-Nearest Neighbors
from sklearn.neighbors import KNeighborsClassifier

# Create the model
knn_model = KNeighborsClassifier()

# Train the model
knn_model.fit(X_train_scaled, y_train)

# Generate predictions
y_pred_knn = knn_model.predict(X_test_scaled)

print("KNN predictions:")
print(y_pred_knn[:10])
```

Why this model is suitable

K-Nearest Neighbors is suitable for this classification problem because it classifies a wine based on the quality classes of similar wine samples. Feature scaling is important for KNN because it uses distances between observations
